# Part 1: Deck synergy, round 2: deck features, nonland synergy, human drafting as a synergy prior, and pooling sets

`archive/deck_synergy.ipynb` found that on MSH, pair synergy didn't replicate across halves of the data, and that when
it was allowed in, it mostly learned the mana base (basics × same-colour cards). Its takeaways listed the next
steps, all run here on **all 12 sets with game data**:

* give the model the mana base directly: **deck features** from Scryfall data (`cardinfo.deck_features`: land
  count buckets, curve shape, number of colours and splash, how well the lands cast the spells)
* fit the FM on **nonland cards only**, with the deck features alongside, so pairs must be something other than
  colour consistency
* **run the split-half test on every set**
* new: **human drafting as a synergy prior.** The pick model (`picknet` `pool`, notebook `drafting.ipynb`) learns
  how much having card j in your pool raises the odds you take card i: S<sub>ij</sub> = ⟨a<sub>i</sub>, e<sub>j</sub>⟩. Drafters
  take cards that go together, so S is a free estimate of "goes together" from 50-200k drafts per set, made
  without looking at a single game result. Two ways to use it: one extra column (the deck's mean pairwise
  affinity, **one fitted weight**), or the FM's pair vectors as a learned linear map of the pick model's card vectors.
* new topology: **a deck transformer shared across all 12 sets**, reading cards only through their features
  (rules text embedding, Scryfall features, card meta), boosted on each set's own linear model (`scripts/archive/deck_multiset.py`)

Every model is a per-game logistic model with the same context as before (play/draw, mulligans, both ranks,
skill and experience buckets, record before the game), on Premier Draft games, with the same chronological split
(test = games of the last 15% of drafts). Scores are **test log loss minus the linear model's**, game by game on the
same games, with 95% intervals (×1000; negative = better). `scripts/deck_synergy_v2.py`.

Sections:

2. Is the synergy real now? Split-half replication on 12 sets
3. Which deck features matter
4. Human drafting affinity: how much it is worth, and what it pairs
5. A deck network shared across sets
6. Takeaways

## 4. Human drafting affinity

**How much is it worth?** `linear + deck features + draft affinity` adds a single column to the deck-features
model, the deck's mean pairwise S<sub>ij</sub> over nonland cards, standardised. Its one weight, per set:

```python
aw = pl.DataFrame([{"set": s,
                    "affinity weight (log-odds per sd)": next(m for m in R[s]["models"] if "affinity weight (log-odds per sd)" in m)["affinity weight (log-odds per sd)"],
                    "gain over deck features x1000": (next(m for m in R[s]["models"] if m["model"] == "linear + deck features + draft affinity")["test logloss"]
                                                    - next(m for m in R[s]["models"] if m["model"] == "linear + deck features")["test logloss"]) * 1000}
                   for s in SETS])
aw.with_columns(pl.exclude("set").round(3))
```

**What does it pair?** The highest-affinity nonland pairs in the set where affinity helped most, among pairs
that share at least 2,000 training decks, with the number of training decks that contain both, and the lowest.

```python
import torch
from deck_synergy_v2 import pick_affinity
from mtgdata.cardinfo import card_info
from mtgdata.games import aligned, game_path
from decks.deckstrength import game_split, load_drafts

best = aw.sort("gain over deck features x1000")["set"][0]
d = load_drafts(best, "PremierDraft")
S, _ = pick_affinity(best, d.names)
g = pl.read_parquet(game_path(best, "PremierDraft"))
sp = game_split(d, g["draft_id"])
X = aligned(g.filter(pl.Series(sp == 0)), "deck_", d.names).astype(np.float32)
nl = ~card_info(d.names, best)["is_land"].to_numpy()
P = (X[:, nl] > 0).astype(np.float32)
co = P.T @ P
names = np.array(d.names)[nl]
Sn = S[np.ix_(nl, nl)]
iu = np.triu_indices(len(names), 1)
ok = co[iu] >= 2000
top = np.argsort(-Sn[iu][ok])[:15]
bot = np.argsort(Sn[iu][ok])[:8]
a, b = iu[0][ok], iu[1][ok]
show = lambda ix: pl.DataFrame({"card a": names[a[ix]], "card b": names[b[ix]], "affinity": Sn[iu][ok][ix].round(2),
                                "decks together": co[iu][ok][ix].astype(int)})
print(best, "- highest human affinity")
display(show(top))
print("lowest")
show(bot)
```

In SOS, the set where affinity is worth most (−1.17 ×1000 on top of deck features), the highest-affinity pairs
are one archetype's package: the Pest cards (Pest Mascot, Bogwater Lumaret, Blech, Teacher's Pest, Essenceknit
Scholar) and the Mascot/Academic pairs. These are the cards drafters take *because* they already have the other
one. The lowest-affinity pairs are cards from different colour pairs that rarely end up in the same pool. MSH is
the one set where the affinity column hurts held-out games (+0.96), even though its weight is positive.

**Is it more than co-occurrence, or colour? And is the draft-level gain just skill?** (`scripts/affinity_checks.py`)
Three single columns of the same form (the deck's mean pairwise score over nonland cards, standardised), each on
top of `linear + deck features`: the pick model's affinity; **deck PMI**, plain co-occurrence of the two cards
in training decks; and **colour overlap**, the cosine of the two cards' colour pips. Game level: test log loss vs
deck features. Draft level: Spearman calibrated with rank only, and with rank + the drafter's skill bucket.

```python
ac = {json.loads(p.read_text())["set"]: json.loads(p.read_text())
      for p in sorted((ROOT / "results").glob("*.PremierDraft/deck_synergy_v2/affinity_checks.json"))}
rows = []
for s, r in ac.items():
    for m in r["models"]:
        rows.append({"set": s, "model": m["model"], "vs DF x1000": m["vs deck features x1000"], "±95%": m["±95% x1000"],
                     "draft rho (rank)": m["draft spearman (rank)"], "draft rho (rank + skill)": m["draft spearman (rank + skill)"]})
acl = pl.DataFrame(rows)
display(acl.group_by("model", maintain_order=True).agg(
    pl.col("vs DF x1000").mean().round(3).alias("game: mean vs DF x1000"),
    (pl.col("vs DF x1000") < 0).sum().alias("sets better"),
    pl.col("draft rho (rank)").mean().round(4), pl.col("draft rho (rank + skill)").mean().round(4)))
w = pl.DataFrame([{"set": s, **next(m for m in r["models"] if m["model"] == "+ all three")["column weights"],
                   "r(pick, PMI)": r["column correlations"]["pick affinity ~ deck PMI"]} for s, r in ac.items()])
w.with_columns(pl.exclude("set").round(3))
```

Pick affinity beats co-occurrence on 10 of 12 sets (LCI ties, MSH is worse for every column) even though the two
columns correlate at r = 0.7-0.84, and when all three go in together, pick affinity takes all the weight (mean
0.18 per sd; PMI −0.01, colour −0.06). So what the pick model adds is not "these cards end up in the same decks" or
"these cards share a colour": it's which cards drafters *take because of* each other. With the skill bucket in
the draft-level calibration, affinity's draft-level gain shrinks from +0.012 to +0.005 Spearman: about 60% of it
was tracking the drafter, the rest survives. (The skill bucket is computed from the same games, so this is a
conservative check.) On MSH every column, even colour overlap, hurts held-out games, which points to a shift
between MSH's training and test weeks rather than to affinity itself.

## 5. A deck network shared across sets

`scripts/archive/deck_multiset.py`. Stage 1 is each set's `linear + deck features` model, fit on that set alone, as a
fixed offset. Stage 2 is one transformer over the deck's distinct cards, shared by all 12 sets, which can only
see a card through its 465 features (no card ids) plus its copies. It's trained on 200k training games per set
(2.4M) to predict what stage 1 misses. Validation is all sets' validation games; test is every test game.
`shared_nodf` is the same without deck features in stage 1, to see whether the shared network finds them itself.

```python
rows = []
for n in ["shared_df", "shared_nodf"]:
    f = ROOT / "results" / "deck_multiset" / f"{n}.json"
    if f.exists():
        r = json.loads(f.read_text())
        for s, v in r["per_set_test"].items():
            rows.append({"run": n, "set": s, "stage 1": v["stage1 logloss"], "shared": v["shared logloss"],
                         "diff x1000": v["diff"] * 1000, "±95%": v["±95%"] * 1000, "set scale": v["set scale"]})
ms = pl.DataFrame(rows)
display(ms.group_by("run").agg(pl.col("diff x1000").mean().round(3).alias("mean diff x1000"),
                               (pl.col("diff x1000") < 0).sum().alias("sets better"), pl.len().alias("sets")))
ms.with_columns(pl.col("stage 1", "shared").round(5), pl.col("diff x1000", "±95%", "set scale").round(3))
```

## 6. Takeaways

1. **Deck features are a free win on every set.** Land-count buckets, curve shape, number of colours and
   splash, and how well the lands cast the spells improve held-out games on 12/12 sets (−0.70 ×1000 on
   average) and draft-level Spearman from 0.230 to 0.249. The biggest effects are about mana: a shortfall of
   coloured sources for the spells (−0.065 log-odds per sd), splashing (−0.04), 3+ main colours (−0.035);
   16 lands beats 17 (+0.035), and a higher average mana value among the nonland cards helps once each card's
   power is held fixed (+0.04).
2. **With the mana base taken out, pair synergy is real on 11 of 12 sets.** Pairs between nonland cards, with
   the deck features alongside, replicate across independent halves of the drafts at r = 0.10-0.31 (card power:
   0.83-0.92) and improve held-out games on 11/12 sets (−1.02). MSH, the smallest set, is the exception on every
   synergy measure (pair r = −0.01), which is why the first synergy notebook, run on MSH, found nothing.
3. **How humans draft is a synergy prior.** The pick model's affinity, learned from picks alone and never from a
   game result, beats the whole nonland FM as a single extra column (−1.10, one fitted weight per set), and the
   FM's learned pairs agree with it (r = 0.09-0.22 where synergy replicates). Stacking both is best on games
   (−1.20), and affinity is best at the draft level (Spearman 0.262 against 0.271 for the draft-trained LightGBM
   deck model and 0.230 for linear). Caveat: at the draft level a coherent build may also mark a stronger
   drafter. On MSH affinity helps the draft-level score (0.248 → 0.276) but not the skill-adjusted game score.
4. **A deck transformer pooled over 12 sets finds general deck structure, but the deck features already have
   it.** Reading cards only through their features, it gains −0.64 ×1000 over plain linear on 12/12 sets
   (`shared_nodf`), about 90% of what the hand-built deck features give (−0.70). On top of linear + deck
   features it adds only −0.09 (9/12). What is left after that is specific to each set (which cards go together),
   and the set's own drafts (affinity) and games (nonland FM) are where it shows up.
5. **It isn't co-occurrence or colour.** Deck co-occurrence (PMI) as the same single column gains less (−0.26 vs
   −0.40 on top of deck features), colour overlap almost nothing (−0.04), and fitted together pick affinity takes
   all the weight. About 60% of its draft-level gain goes away once the drafter's skill is in the calibration;
   the game-level gain is skill-adjusted already.
6. **Next:** use the affinity as a prior *inside* the FM (shrink pair vectors toward the affinity, not toward
   zero); try affinity from PickNet (context-aware) instead of the bilinear pool model.

# Part 2: Deck synergy, rounds 3–7: training, stacking, and synergy from cards seen together

`synergy.ipynb` ended with the nonland FM + deck features + the pick model's affinity column as the best
game-level deck model (−1.20 ×1000 test log loss vs linear), and 0.262 draft-level Spearman against 0.271 for the
draft-trained LightGBM deck model. This notebook covers the five rounds after it, on the same 12 sets:

| round | script | question |
|---|---|---|
| 3 | `deck_synergy_v3.py`, `deck_synergy_v3_blend.py` | more from affinity (profile, per-card fit)? does the game model add to LightGBM? |
| 4 | `deck_synergy_v4.py`, `deck_synergy_v4_draft.py` | round 2's pair variants, properly trained; LightGBM with deck features, boosted from the game model |
| 5 | `deck_synergy_v5.py` | learn pairs from the cards each game actually *saw*, not the decklist |
| 6 | `deck_synergy_v6.py` | is that seen-together synergy real? held-out test and an unseen-card placebo |
| 7 | `deck_synergy_v7.py` | final pair tables and per-card synergy hubs |

Every game-level number is test-game log loss, paired against a reference model on the same games, ×1000 (lower is
better), with the 95% interval of the paired difference. Draft-level numbers are Spearman of the calibrated deck
score against the test drafts' W/(W+L).

```python
import json
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import polars as pl

ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent

BLUE, ORANGE, AQUA, YELLOW, MAGENTA, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#e87ba4", "#4a3aa7"
GRAY, INK, MUTED, SURFACE, GRID = "#a3a29c", "#0b0b0b", "#52514e", "#fcfcfb", "#e6e5e0"
plt.rcParams.update({
    "figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
    "axes.edgecolor": MUTED, "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.8, "axes.axisbelow": True,
    "lines.linewidth": 2, "font.size": 10, "legend.frameon": False,
})
pl.Config.set_tbl_rows(40); pl.Config.set_fmt_str_lengths(60); pl.Config.set_tbl_width_chars(220)


def load(round_dir, name="results.json"):
    return {p.parts[-3].split(".")[0]: json.loads(p.read_text())
            for p in sorted((ROOT / "results").glob(f"*.PremierDraft/{round_dir}/{name}"))}


def by_model(res):
    return {s: {m["model"]: m for m in r["models"]} for s, r in res.items()}


def dots(ax, series, key, err, ref_label):
    """Per-set dots with 95% intervals, one colour per series in fixed order."""
    cols = [BLUE, ORANGE, AQUA, YELLOW, VIOLET, MAGENTA]
    for j, (label, per_set) in enumerate(series.items()):
        x = np.arange(len(SETS)) + (j - (len(series) - 1) / 2) * 0.16
        ax.errorbar(x, [per_set[s][key] * 1000 for s in SETS], yerr=[per_set[s][err] * 1000 for s in SETS],
                    fmt="o", ms=4.5, color=cols[j], label=label, elinewidth=1)
    ax.axhline(0, color=MUTED, lw=1)
    ax.set_xticks(np.arange(len(SETS)), SETS)
    ax.set_ylabel(f"test log loss vs {ref_label} (×1000)")
    ax.legend(fontsize=8, loc="lower left")


V3, V4, V5, V6, V7 = (by_model(load(f"deck_synergy_v{i}")) for i in (3, 4, 5, 6, 7))
SETS = sorted(V3)
print(f"{len(SETS)} sets: {', '.join(SETS)}")
```

## 1. Round 3: the earlier models were under-trained

Round 3 tried two ways to get more out of the pick model's affinity than one mean-affinity column:

* **affinity profile:** the fit distribution, not just its mean: the worst-fitting card, the spread of fit, the
  count of off-plan cards (below the training 10th percentile of fit)
* **card fit:** each card gets its own weight on how well it fits the rest of the deck (a build-around vs a card
  that is good anywhere), warm-started from the nested model

Card fit first looked like a 12/12 win. The control explains why: the *same* warm-started continuation with card fit
held at zero did just as well. Round 2's models stopped at their first validation plateau at learning rate 3e-3.
`synergy._train` now takes `lr_drops` (on a plateau, return to the best state and divide the rate by 3), and every
model from here on trains with two drops. The chart below is the final run with the fix: against the properly trained
FM + affinity, card fit and the continuation control are both close to zero.

```python
ref = "fm4 nonland + deck features + aff"
series = {"card fit": {s: V3[s][ref + " + card fit"] for s in SETS},
          "continued training, no card fit (control)": {s: V3[s][ref + ", continued (control)"] for s in SETS},
          "affinity profile (linear)": {s: {k: V3[s]["linear + deck features + affinity profile"][k]
                                             - V3[s]["linear + deck features + aff"][k] * (k == "vs v2 best")
                                             for k in ("vs v2 best", "vs v2 best ±95%")} for s in SETS}}
series["affinity profile (linear)"] = {s: {"vs v2 best": V3[s]["linear + deck features + affinity profile"]["vs v2 best"]
                                                         - V3[s]["linear + deck features + aff"]["vs v2 best"],
                                           "vs v2 best ±95%": V3[s]["linear + deck features + affinity profile"]["vs v2 best ±95%"]}
                                       for s in SETS}
fig, ax = plt.subplots(figsize=(11, 3.8))
dots(ax, series, "vs v2 best", "vs v2 best ±95%", "the nested model")
ax.set_title("round 3 additions vs the model they extend (lower is better)", loc="left")
plt.tight_layout()

t = pl.DataFrame([{"model": m, "mean vs linear ×1000": np.mean([V3[s][m]["vs linear"] for s in SETS]) * 1000,
                   "mean draft Spearman": np.mean([V3[s][m]["draft spearman"] for s in SETS])}
                  for m in V3["MSH"]])
t.with_columns(pl.exclude("model").round(3))
```

The "affinity profile" dots are the profile model minus the single-column model; the profile's interval is against
the FM, so it is only a rough guide. **The mean carries all the affinity signal:** fitted together, the mean gets
+0.13 log-odds per sd on 12/12 sets, while the worst card, spread and off-plan count average +0.01 to +0.03.

## 5. Round 7: synergy hubs and pairs

Round 7 refits the seen FM with the penalty grid extended down to 3e-5; round 6's grid had bottomed out on 11 of 12
sets. The held-out gain is unchanged (−2.43 ×1000) and the grid now brackets the optimum. From the final fits:

* `seen_pairs.csv`: Q<sub>ij</sub>, the extra log-odds when both cards are seen, beyond each card's own seen value
* `card_hubs.csv`: each card's own seen value u and its **synergy dependence**, the co-occurrence-weighted mean Q
  over its partners

A synergy hub is a card whose value depends on what it is drawn with. `card_hubs.csv`'s *synergy dependence*
(mean Q over partners) has a catch: a card's own value u and its average pair effect trade off against each other
(r = −0.55 to −0.69 across cards), so a weak card can look synergy-dependent partly by construction. The
**partner spread**, the co-occurrence-weighted sd of Q across a card's partners, can't be faked that way (r with u:
−0.23 to −0.37). It asks whether the card is much better with some partners than with others. Below, cards with at
least 20 partners; the labelled ones are each set's top 4 by spread, with their best partner.

```python
def hubs(s, min_partners=20):
    base = ROOT / "results" / f"{s}.PremierDraft" / "deck_synergy_v7"
    p = pl.read_csv(base / "seen_pairs.csv")
    both = pl.concat([p.select(pl.col("card a").alias("card"), "seen Q", "train decks"),
                      p.select(pl.col("card b").alias("card"), "seen Q", "train decks")])
    w = pl.col("train decks")
    mean = (pl.col("seen Q") * w).sum() / w.sum()
    spread = both.group_by("card").agg(
        (((pl.col("seen Q") - mean) ** 2 * w).sum() / w.sum()).sqrt().alias("partner spread"),
        pl.len().alias("partners"))
    return pl.read_csv(base / "card_hubs.csv").join(spread, on="card").filter(pl.col("partners") >= min_partners)


def repel(ys, gap):
    """Label y positions, top-down, at least `gap` apart."""
    out = []
    for y in ys:
        out.append(min(y, out[-1] - gap) if out else y)
    return out


fig, axes = plt.subplots(1, 3, figsize=(13, 4.4), sharey=True)
for ax, s in zip(axes, ["BLB", "MKM", "SOS"]):
    h = hubs(s)
    ax.scatter(h["seen value u"], h["partner spread"], s=10, color=GRAY, alpha=0.6)
    top = h.sort("partner spread", descending=True).head(4)
    ax.scatter(top["seen value u"], top["partner spread"], s=24, color=ORANGE, edgecolor=SURFACE, linewidth=1)
    x_lab = h["seen value u"].max() + 0.05
    for r, y in zip(top.iter_rows(named=True), repel(top["partner spread"].to_list(), 0.0045)):
        ax.annotate(f"{r['card']}  (+ {r['best partner']})", (r["seen value u"], r["partner spread"]),
                    xytext=(x_lab, y), fontsize=7.5, color=INK, va="center",
                    arrowprops=dict(arrowstyle="-", color=MUTED, lw=0.6))
    ax.set_title(s, loc="left"); ax.set_xlabel("own value when seen (log-odds)")
    ax.set_xlim(right=x_lab + 0.9)
    ax.set_xticks([-0.5, 0, 0.5])
axes[0].set_ylabel("partner spread (sd of Q over partners)")
fig.suptitle("synergy hubs: cards that are much better with some partners than others", x=0.01, ha="left")
plt.tight_layout()
```

```python
pl.concat([hubs(s).sort("partner spread", descending=True).head(4).with_columns(pl.lit(s).alias("set")) for s in SETS]) \
  .select("set", "card", "seen value u", "partner spread", "synergy dependence", "best partner", "best Q", "partners") \
  .with_columns(pl.col("seen value u", "partner spread", "synergy dependence", "best Q").round(3))
```

The strongest pairs per set (both cards in ≥ 2,000 training decks), with the pick model's affinity alongside. The
two mostly disagree (r ≈ 0.07): drafters group cards by archetype, while these pairs are about cards that are
better when drawn together.

```python
def top_pairs(s, n=5):
    p = pl.read_csv(ROOT / "results" / f"{s}.PremierDraft" / "deck_synergy_v7" / "seen_pairs.csv")
    return p.head(n).with_columns(pl.lit(s).alias("set"))


pl.concat([top_pairs(s) for s in SETS]).select("set", "card a", "card b", "seen Q", "pick affinity", "train decks") \
  .with_columns(pl.col("seen Q", "pick affinity").round(3))
```

## 6. Takeaways

1. **Train properly before comparing.** Round 2's models stopped at their first plateau. With learning-rate drops,
   the nonland FM + affinity goes from −1.20 to −1.40 ×1000 vs linear (linear retrained the same way), and the
   ranking of models doesn't change. A continued-training control is what showed that card fit's apparent gain
   was just extra training.
2. **Deck-level synergy is saturated.** Every pair-term variant gains about −0.15 to −0.18 ×1000 over the affinity
   column. Rank, lands in the pairs, and pick-model-derived pairs make no difference; neither do affinity profiles
   or per-card fit.
3. **The best deck-strength model is a blend.** LightGBM (with deck features + affinity) averaged with the
   skill-adjusted game FM: 0.279 mean Spearman, against 0.271 for LightGBM and 0.244 for the linear deck model.
   Each model gets something the other misses, and further stacking moves it by 0.001 or less.
4. **Synergy looked much larger in the game than in the decklist, but that was mostly card draw** (Part 3). Pairs of
   cards seen together improved held-out games by −2.4 ×1000 and replicated at r ≈ 0.5, but "seen" includes cards
   drawn by other cards. On the opening hand, partner-specific synergy is −0.27 ×1000.
5. **Why that doesn't improve deck strength:** a deck only collects a pair's effect in the ~10% of games where both
   cards show up, and the deck-level average is already in the FM and affinity. Deck-strength prediction is now
   limited by Bo1 noise, not by its synergy estimates.
6. **Next** (as of round 7; see Part 3): use the seen-together pair values where synergy drives a decision, as a synergy term in the pick bots
   and table simulator; with replay data for more sets, look at pairs that were actually cast together and at
   your-card × opponent-card matchups.

# Part 3: Deck synergy, rounds 8–9: is the in-game synergy causal, and does it carry across sets?

Results tables only (`scripts/deck_synergy_v8.py`, `scripts/deck_synergy_v9.py`; per-set JSON in
`results/<SET>/deck_synergy_v8/` and `results/deck_synergy_v9/`). Test-game log loss ×1000, mean over 12 sets.

## 1. Round 8: opening hand vs cards seen

Round 6's seen-together pairs had two problems. Most of their effect wasn't partner-specific: pushing each card's mean
pair value to zero removed ~70% of the gain on SOS. And "seen" includes cards drawn by other cards' effects, so it is
partly an outcome of play. The opening hand is dealt before play (mulligans are in the context), so it is the clean
input. A per-card *slope* on how many other spells are in hand / were seen models the non-specific part.

| model | opening hand | seen this game |
|---|---|---|
| per-card slope, vs no pairs | −0.13 (2/12 significant) | −4.53 (12/12) |
| free pairs (rank 4), vs no pairs | −0.30 (8/12) | −2.44 (12/12) |
| **slope + pairs, vs slope** (partner-specific) | **−0.27** (better on 11/12, 8 significant) | −0.59 (10/12) |
| split-half pair r | 0.13 | 0.46 free / 0.21 on top of slope |

On seen cards the top slopes are draw and recursion spells (SOS: Glimpse of Nature, Wisdom of Ages, Decorum
Dissertation): a resolved draw spell both shows you more cards and wins games. The clean opening-hand synergy is real
but small, about the size of deck-level pairs (−0.18), with similar replication. Top clean pairs: SOS's Potioner's
Trove with Snarl Song and Together as One; Garrison Excavator with Rubble Rouser.

## 2. Round 9: opening-hand synergy from card features, across sets

Stage 1 is each set's opening-hand model without pairs (a fixed offset). Stage 2 is one pair model shared by all 12
sets that reads cards only through their features (rules-text embedding, metadata, Scryfall facts), with a per-set scale.

| pair model | mean vs no pairs | sets better (significant) |
|---|---|---|
| per-set free pairs (round 8) | −0.30 | 11/12 (8) |
| shared, from card features | −0.20 | 11/12 (6) |
| shared, set held out (day zero): DFT / BLB / LCI | +0.75 / +0.26 / −0.24 | 1/3 |

Card features carry part of the synergy (shared pairs correlate 0.1–0.45 with each set's own), not most. For a set it

text fails.

## 3. Takeaways

1. **Round 6's big in-game "synergy" was mostly card draw.** A per-card slope on spells seen takes −4.5 of the
   −5.1 gain on seen cards; partner-specific pairs on top add −0.6, and part of that is likely draw too.
2. **Clean synergy is real but small.** On opening hands, partner-specific pairs gain −0.27 ×1000 (11/12 sets),
   comparable to deck-level pairs, and replicate at r ≈ 0.13.
3. **It has to be learned within each set.** Card features explain part of it, but don't predict synergy on a new set.
4. **Next:** replay data for more sets (only MSH has it here): pairs actually cast together and when, and
   your-card × opponent-card matchups.
