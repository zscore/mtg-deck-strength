## 1. The target: an event record

A Premier Draft run is best-of-one games until **7 wins or 3 losses**, so every finished run is 7-x or x-3. Some
runs are abandoned early (0-0, 2-1, ...). The model's output is one number per drafted deck, **p = the chance of
winning a game**.

The run's stopping rule depends only on the results so far, so it drops out of the likelihood: a run that went
W-L has likelihood p<sup>W</sup>(1-p)<sup>L</sup>, whatever order the games came in and whether the run finished.
Every model below is trained on this binomial loss. In practice that is a logistic regression in which each draft
appears twice: as a "win" row weighted W and a "loss" row weighted L. Abandoned runs count for exactly as many
games as they played. Only the 0-0 runs carry no information.

```python
fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 3.8), gridspec_kw={"width_ratios": [1.15, 1]})
grid = np.zeros((4, 8), dtype=int)
np.add.at(grid, (l.astype(int), w.astype(int)), 1)
im = a1.imshow(grid, cmap="Blues", aspect="auto", vmin=0, vmax=grid.max() * 1.15)
for (li, wi), n in np.ndenumerate(grid):
    if n:
        a1.text(wi, li, f"{n:,}", ha="center", va="center", fontsize=8,
                color="white" if n > grid.max() * 0.55 else INK)
a1.set_xticks(range(8)); a1.set_yticks(range(4)); a1.grid(False)
a1.set_xlabel("wins"); a1.set_ylabel("losses")
a1.set_title("Every draft's record (all splits)", loc="left")

ok = (w + l) > 0
a2.hist((w / np.maximum(w + l, 1))[ok], bins=np.linspace(0, 1, 23), color=BLUE, rwidth=0.9)
a2.set_xlabel("game win rate of the run, W / (W + L)"); a2.set_ylabel("drafts")
a2.set_title(f"Pooled game win rate {w.sum() / (w + l).sum():.1%}", loc="left")
plt.tight_layout(); plt.show()
```

The pooled win rate is above 50% because 17Lands users are stronger than the average Arena player they get paired
against.

## 2. Scoring, and the noise ceiling

Models are compared on held-out **test drafts**, the last 15% by draft time, with the same rows for every model.
There are two metrics:

* **Spearman** rank correlation between predicted p and the run's observed W/(W+L). This is the headline metric of
  the DraftEncoder paper (arXiv 2607.04782), which these experiments follow.
* **NLL per game**: the binomial log loss above, divided by games played. It rewards calibration as well as ranking.
  Always guessing the pooled rate scores about 0.688.

Neither number can get close to its ideal. A run is at most 9 Bo1 games, so even a deck with a true 60% win rate
often goes 2-3 or 7-1. The **ceiling** below asks how well a perfect model could score: draw each deck's true p,
play out its run under the 7-win / 3-loss rule, and compute Spearman(true p, simulated W/(W+L)). The answer depends
only on how much true deck strength varies.

```python
rng = np.random.default_rng(0)
sds = np.linspace(0.01, 0.15, 15)
ceil = []
for sd in sds:
    p_true = np.clip(rng.normal(w.sum() / (w + l).sum(), sd, 40_000), 0.02, 0.98)
    ceil.append(spearmanr(p_true, simulate_runs(p_true, seed=1)).statistic)
fig, ax = plt.subplots(figsize=(7.5, 3.6))
ax.plot(sds * 100, ceil, color=BLUE, marker="o", ms=4)
ax.set_xlabel("spread of true deck win rates across drafts (sd, percentage points)")
ax.set_ylabel("best possible Spearman")
ax.set_ylim(0, 0.8)
ax.set_title("Even a perfect model scores well below 1: each run is a handful of Bo1 games", loc="left")
plt.tight_layout(); plt.show()
print("ceiling at sd = 3, 5, 8 points:", [round(float(np.interp(x, sds, ceil)), 3) for x in (0.03, 0.05, 0.08)])
```

For each fitted model, `deck_metrics` also reports the ceiling *for that model's own predicted spread*: the score
it would get if its predictions were exactly right. When a model's test Spearman is close to its ceiling, it is
about as good as it could be at the spread it predicts. To score higher it has to find real differences between
decks that it currently misses.

## 3. The models and what each one sees

| family | input | models |
|---|---|---|
| baseline | player rank only | `rank_only` |
| heuristics | final pool or deck, scored with public card win rates | `pool_top23:gih_wr` (mean GIH WR of the best 23 cards), `deck_mean:gih_wr` |
| linear | final 42-card **pool**, as card counts | `linear_pool` |
| linear | registered 40-card **deck** (lands included), as card counts | `linear_deck`, `+structure`, `+archetype`, `+gih_prior` |
| trees | deck counts + structure features | `lgbm_deck+structure` |
| game-level | every game's deck, adjusted for play/draw, mulligans, skill and both ranks | `game_fm` (card power + pair synergy) |
| transformer | the whole **pick sequence** (every pack seen and the card taken), optionally + the deck | `draftencoder_{paper,big}[+deck]` |

The registered deck comes from 17Lands *game* data: the build the drafter actually played most. Any model that
sees the deck also benefits from the drafter's own build decisions, such as which 23 cards to play and how many
lands. The pool and pick-sequence models must work those out themselves. All models also get the player's rank;
none get the 17Lands skill bucket (the user's overall game win rate), because that is close to the answer.

## 4. The linear deck model, fitted live

`linear_deck` is a binomial logistic regression on the registered deck's card counts plus a one-hot of rank:

logit p = b + Σ<sub>cards</sub> weight<sub>c</sub> · copies<sub>c</sub> + rank<sub>r</sub>

It is simple, fits in seconds, and is within a hair of the best model on most sets. The L2 strength `C=0.01` is
the value chosen on the validation drafts in `train_draftencoder.py`.

```python
ranks = np.eye(N_RANKS, dtype=np.float32)[d.rank]
dk = deck.astype(np.float32)
X_lin = sparse.hstack([sparse.csr_matrix(dk), sparse.csr_matrix(ranks)]).tocsr()
lin = binomial_logreg(X_lin[tr], w[tr], l[tr], C=0.01)
p_lin = lin.predict_proba(X_lin)[:, 1]
m_lin = deck_metrics(p_lin[te], w[te], l[te])
print(f"linear_deck on {m_lin['n']:,} test drafts: Spearman {m_lin['spearman']:.4f}, NLL {m_lin['nll']:.4f}, "
      f"own ceiling {m_lin['ceiling']:.3f}")
print(f"predicted p on test: mean {p_lin[te].mean():.3f}, sd {p_lin[te].std():.3f}, "
      f"5th-95th pct {np.percentile(p_lin[te], 5):.3f}-{np.percentile(p_lin[te], 95):.3f}")
```

The predictions spread out with an sd of about 6 percentage points, and at that spread the noise ceiling is about
0.26. The model's actual Spearman is right at that level. So linear_deck is not wasting the spread it predicts;
**a better model has to find larger real differences between decks**, not rank the same differences more cleanly.

**Calibration.** Bin the test drafts by predicted p and compare each bin's mean prediction with the bin's pooled
win rate, ΣW / Σ(W+L). Points on the diagonal mean "p = 0.60" really does win about 60% of games.

```python
def calib(p, w, l, bins=10):
    edges = np.quantile(p, np.linspace(0, 1, bins + 1))
    b = np.clip(np.searchsorted(edges, p, side="right") - 1, 0, bins - 1)
    return (np.array([p[b == i].mean() for i in range(bins)]),
            np.array([w[b == i].sum() / (w[b == i] + l[b == i]).sum() for i in range(bins)]),
            np.array([(w[b == i] + l[b == i]).sum() for i in range(bins)]))

pm, obs, n = calib(p_lin[te], w[te], l[te])
fig, ax = plt.subplots(figsize=(5.2, 4.4))
lo, hi = min(pm.min(), obs.min()) - 0.02, max(pm.max(), obs.max()) + 0.02
ax.plot([lo, hi], [lo, hi], color=GRAY, lw=1.2, ls="--")
se = np.sqrt(obs * (1 - obs) / n)
ax.errorbar(pm, obs, yerr=1.96 * se, color=BLUE, marker="o", ms=6, capsize=0, lw=2)
ax.set_xlabel("predicted p (decile mean)"); ax.set_ylabel("observed game win rate")
ax.set_xlim(lo, hi); ax.set_ylim(lo, hi)
ax.set_title("linear_deck calibration, test drafts", loc="left")
plt.tight_layout(); plt.show()
print(f"worst-decile decks win {obs[0]:.1%} of games, best-decile decks {obs[-1]:.1%}")
```

**Card weights vs public win rates.** Each card's weight is the change in log-odds per copy in the deck. The
17Lands GIH WR (win rate when the card is in hand) measures much the same thing, so the two should broadly
agree. The weights are fitted jointly, so a card that is good only because it shows up in strong decks is
credited less here than in GIH WR. Only cards in at least 300 training decks are shown.

```python
C = len(names)
wt = lin.coef_[0, :C]
n_decks = (deck[tr] > 0).sum(0)
cards = pl.DataFrame({"card": names, "weight": wt, "gih_wr": gih, "train_decks": n_decks})
common = cards.filter(pl.col("train_decks") >= 300)

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(common["gih_wr"] * 100, common["weight"], s=12, color=BLUE, alpha=0.6, lw=0)
r = spearmanr(common["gih_wr"], common["weight"]).statistic
lab = pl.concat([common.sort("weight").head(4), common.sort("weight").tail(3)])
for row in lab.iter_rows(named=True):
    ax.annotate(row["card"], (row["gih_wr"] * 100, row["weight"]), fontsize=7.5, color=MUTED,
                xytext=(4, 2), textcoords="offset points")
ax.axhline(0, color=GRAY, lw=1)
ax.set_xlabel("GIH WR from train games (%)"); ax.set_ylabel("linear_deck weight (log-odds per copy)")
ax.set_title(f"{common.height} cards in ≥300 training decks: Spearman(GIH WR, weight) = {r:.2f}", loc="left")
plt.tight_layout(); plt.show()
```

```python
show = lambda df: df.with_columns(pl.col("weight").round(3), (pl.col("gih_wr") * 100).round(1))
print("Highest weights"); display(show(common.sort("weight", descending=True).head(10)))
print("Lowest weights"); display(show(common.sort("weight").head(10)))
```

**Rank.** Arena pairs players within their rank, so rank decides *who you play*. With the deck held fixed,
the same 40 cards win less often at Diamond or Mythic than at Bronze, because the opponents are better. Rank is
not a proxy for the drafter's skill here; that would push the other way. Shown relative to Bronze.

```python
rk = lin.coef_[0, C:]
labels = ["unknown"] + list(RANK_ORDER)
base = rk[1]
pmean = w[tr].sum() / (w[tr] + l[tr]).sum()
fig, ax = plt.subplots(figsize=(6.5, 2.8))
vals = rk - base
ax.bar(labels, vals, color=[GRAY] + [BLUE] * 6, width=0.6)
for i, v in enumerate(vals):
    if i == 1: continue
    ax.text(i, v + (0.004 if v >= 0 else -0.004), f"{v:+.2f}", ha="center",
            va="bottom" if v >= 0 else "top", fontsize=8.5, color=INK)
ax.axhline(0, color=MUTED, lw=1)
ax.set_ylabel("log-odds vs Bronze"); ax.grid(axis="x", visible=False)
ax.set_title("Rank effect at a fixed deck: higher rank, tougher opponents", loc="left")
plt.tight_layout(); plt.show()
```

**A worked example.** The highest- and lowest-rated test decks, with each card's contribution to the logit
(weight × copies). Basic lands have weights too: once the deck size is fixed, a basic land's weight says how
playing it compares with playing a spell.

```python
def explain(i, k=8):
    row = deck[i]
    idx = np.flatnonzero(row)
    contrib = wt[idx] * row[idx]
    df = (pl.DataFrame({"card": names[idx], "copies": row[idx].astype(int), "contribution": contrib})
          .sort("contribution", descending=True).with_columns(pl.col("contribution").round(3)))
    print(f"predicted p = {p_lin[i]:.3f}; actual record {int(w[i])}-{int(l[i])}; rank {labels[d.rank[i]]}; "
          f"card total {contrib.sum():+.2f} log-odds over {int(row.sum())} cards")
    display(pl.concat([df.head(k), df.tail(4)]))

te_idx = np.flatnonzero(te)
explain(te_idx[np.argmax(p_lin[te])])
explain(te_idx[np.argmin(p_lin[te])])
```

## 5. Deck structure, and LightGBM

A sum of per-card weights cannot see *how the deck is built*: 15 lands vs 18, a top-heavy curve, a three-colour
mana base. `structure_features` (in `improve_deck.py`) adds:

* land count, one-hot from 14 to 19+
* the nonland curve: card counts at mana values 0-1, 2, 3, 4, 5, 6+
* creature count
* number of main colours (≥4 nonland cards of that colour), and number of splashed colours

Below, `+structure` is refitted live and its structure coefficients are plotted. Land and colour counts are
one-hot, so read those relative to each other. Curve and creature terms are per 5 and per 10 cards.

```python
st = structure_features(deck, d.names, SET)
st_names = ([f"{n} lands" for n in ("≤14", "15", "16", "17", "18", "≥19")]
            + [f"curve MV {m} (per 5)" for m in ("0-1", "2", "3", "4", "5", "6+")]
            + ["creatures (per 10)"] + [f"{n} main colour{'s' if n > 1 else ''}" for n in (1, 2, 3, 4)]
            + ["splash colours"])
X_st = sparse.hstack([X_lin, sparse.csr_matrix(st)]).tocsr()
lst = binomial_logreg(X_st[tr], w[tr], l[tr], C=0.01)
p_st = lst.predict_proba(X_st)[:, 1]
m_st = deck_metrics(p_st[te], w[te], l[te])
print(f"+structure: Spearman {m_st['spearman']:.4f} (linear_deck {m_lin['spearman']:.4f}), "
      f"NLL {m_st['nll']:.4f} (linear_deck {m_lin['nll']:.4f})")

coef = lst.coef_[0, -st.shape[1]:]
groups = [(0, 6, BLUE), (6, 12, ORANGE), (12, 13, AQUA), (13, 17, YELLOW), (17, 18, MAGENTA)]
fig, ax = plt.subplots(figsize=(8, 5.6))
y = np.arange(len(coef))[::-1]
for a, b, c in groups:
    ax.barh(y[a:b], coef[a:b], color=c, height=0.65)
for yi, v in zip(y, coef):
    ax.text(v + (0.004 if v >= 0 else -0.004), yi, f"{v:+.3f}", va="center",
            ha="left" if v >= 0 else "right", fontsize=8, color=INK)
ax.set_yticks(y, st_names); ax.axvline(0, color=MUTED, lw=1); ax.grid(axis="y", visible=False)
ax.set_xlabel("log-odds"); ax.margins(x=0.2)
ax.set_title("+structure: coefficients on top of the card weights", loc="left")
plt.tight_layout(); plt.show()
lands, cnt = np.unique(st[has_deck, :6].argmax(1), return_counts=True)
print("share of decks by land count:", {st_names[i].split()[0]: f"{c / cnt.sum():.1%}" for i, c in zip(lands, cnt)})
```

How to read this:

* **Colours matter most.** Holding the cards fixed, each extra main colour costs about 0.1-0.15 log-odds, and each
  splashed colour costs another 0.06. That is the price of an inconsistent mana base, which a per-card sum cannot see.
* **More creatures help** (+0.07 per 10).
* **Land count and curve terms are small and noisy.** About 76% of decks play 17 lands and 22% play 16, so the
  other land counts have little data. The 16 vs 17 difference (+0.06 for 16) suggests 16-land decks were slightly
  better than their cards alone predict on MSH. That could be a real effect, or it could be that the players who
  cut a land are better players.

Together, these features add about +0.012 Spearman on MSH and +0.015 on average across sets.

**LightGBM** (`lgbm_deck+structure`) gets the same inputs (deck counts, structure, rank) and can learn
interactions, such as "this card is only good with enough creatures". It uses the same two-rows-per-draft binomial
trick, with early stopping on validation drafts. It is not refitted here: LightGBM and PyTorch each bundle their
own OpenMP runtime, and loading both in one process crashes on macOS (see the comment in `improve_deck.py`). The
saved results:

```python
imp = json.loads((OUT / "draftencoder" / "improve_deck.json").read_text())
abl = json.loads((OUT / "draftencoder" / "lgbm_ablation.json").read_text())
tab = lambda rows: pl.DataFrame([{"model": k, "Spearman": round(v["spearman"], 4), "NLL/game": round(v["nll"], 5),
                                  "own ceiling": round(v["ceiling"], 3)} for k, v in rows.items()])
print(f"improve_deck.py, {SET} test drafts with a deck"); display(tab(imp))
print("LightGBM ablation: which inputs matter"); display(tab(abl))
```

LightGBM and `+structure` tie on MSH. Averaged across sets, LightGBM is ahead by about +0.013 (section 8). The
ablation shows where its signal comes from: **structure alone is weak** (0.09), and deck counts plus rank already
reach 0.28. Counts beat presence: a second copy of a good card adds value. The best MSH number, 0.285, is a logit
average of `+structure` and LightGBM. The two make different enough errors that averaging helps a little.

## 6. Game-level models: card power, pair synergy, and cards cast

The draft-level models above learn from about 40k records. 17Lands game data has one row per **game**, about 8×
as many examples. Each game also records context the record does not: play/draw, both players' mulligans, the
user's skill bucket, and both players' ranks. `deck_synergy.py` fits

P(win game) = σ(b + f(deck) + covariates),   f(x) = w·x + Σ<sub>i<j</sub> x<sub>i</sub>x<sub>j</sub>⟨v<sub>i</sub>, v<sub>j</sub>⟩

The covariates absorb skill, so w is *skill-adjusted card power*. The second term is a factorization machine:
each card gets a small vector v, and a pair's synergy is the dot product of their vectors (rank 4 or 16). f is then
used as a deck score for the test records like any other model (`game_fm`).

```python
syn = json.loads((OUT / "deck_synergy" / "results.json").read_text())
display(pl.DataFrame([{"model": k, "game log loss": round(v["test_logloss"], 4), "game AUC": round(v["test_auc"], 4),
                       "draft Spearman": round(v["draft_spearman"], 4) if "draft_spearman" in v else None}
                      for k, v in syn.items()]))
power = pl.read_csv(OUT / "deck_synergy" / "card_power.csv")
pairs = pl.read_csv(OUT / "deck_synergy" / "pair_synergy.csv")
print(f"sd of linear card power: {power['power_linear'].std():.3f} log-odds; "
      f"sd of pair synergy (pairs in ≥2000 training games): {pairs['synergy'].std():.4f}")
print("Strongest positive pairs"); display(pairs.sort("synergy", descending=True).head(8)
                                           .with_columns(pl.col("synergy").round(4)))
print("Strongest negative pairs"); display(pairs.sort("synergy").head(8).with_columns(pl.col("synergy").round(4)))
```

Two results:

* **Pair synergy is almost undetectable.** Pair terms have an sd of about 0.001 log-odds, against 0.06 for single
  cards, and the FMs do not beat the linear model on held-out games. The top pairs make thematic sense (the HYDRA /
  artifact package, the Wakandan / Birnin Zana pair), but they are tiny effects. At this data size, a deck's
  strength is close to the sum of its cards.
* **More data per example did not help.** Fitted per game with skill controls, the card power scores 0.251 as a
  deck score, *below* the draft-level `linear_deck` (0.268). Controlling for the skill bucket also removes signal
  the record target can use: good players build better decks *and* win more with them. The draft-level model is
  allowed to credit the deck for both.

**Cards cast are a trap.** Replay data records every spell each player cast, turn by turn. A Bradley-Terry fit on
(user casts − opponent casts) predicts single games extremely well, as the AUC climbs with the turn cutoff below.
Most of that is reverse causation, though: the player who is winning gets to cast more spells. Turned back into a
card value and used to score decks, it is *worse* than plain GIH WR (`train_deckstrength.py`):

```python
ds = json.loads((OUT / "deckstrength" / "results.json").read_text())
display(pl.DataFrame([{"game model": k, "game AUC": round(v["auc"], 4)} for k, v in ds["game"].items()]))
keys = ["pool_top23:gih_wr", "pool_top23:cast_bt_t3", "pool_top23:cast_bt_t8", "pool_top23:cast_bt_t30",
        "pool_top23:gp_wr"]
display(pl.DataFrame([{"deck score (top-23 of the pool)": k.split(":")[1], "draft Spearman": round(ds["deck"][k]["spearman"], 4)}
                      for k in keys]))
```

## 7. DraftEncoder: reading the whole draft

DraftEncoder (arXiv 2607.04782) predicts the run's record from the **pick sequence**: all 42 packs the drafter saw
and the card taken from each. It outputs a prediction after *every* pick. The paper builds card features from
oracle text; this reimplementation (`src/decks/draftencoder.py`) has no oracle text, so a card is a learned
embedding plus metadata (public win rates, colours, mana value, creature, rarity).

* **Set context.** Every card in the set is a token. A FiLM layer conditions them on player rank, then
  self-attention runs over the whole set, giving card embeddings in the context of this set and rank.
* **Draft layers** (5 in `paper`, 8 in `big`). At each pick t there are tokens for the pack's cards and one
  *state* token s[t], initialised with the picked card. In each layer, the pack cards attend causally (with RoPE)
  over the states s[0..t]: what has the drafter committed to? Then s[t] attends over its pack, with the picked
  card as the query: what did they take, given what else was there?
* **Head.** p(win) from [s[t], rank] after each pick. The binomial loss is weighted (t+1)/42, so later picks count more.
* **Deck stage** (`+deck`, not in the paper). The registered deck as (card, copies) tokens plus the final state,
  two or three self-attention layers, and a second head.

## 9. Takeaways

1. **The registered deck is the best input.** On every one of the 12 sets, `linear_deck` beats `linear_pool` by
   about 0.03 Spearman: which 23 cards and which lands the drafter chose carries real information. The pick
   sequence adds nothing on top of it.
2. **The best model is simple.** Deck counts + rank + a few structure features (colours, splash, creatures), fitted as
   a binomial logistic regression or LightGBM, scores 0.24-0.29 Spearman across sets. LightGBM leads on average
   (0.271 vs 0.259) and ties on MSH, where an average of the two is best (0.285).
3. **That is close to the ceiling for the spread these models predict.** Bo1 runs of at most 9 games are noisy, so
   even a perfect model, at the spread these models predict (sd about 6 points), would score only about 0.26-0.28. Further progress has
   to come from finding bigger real differences between decks.
4. **The transformer doesn't pay off here.** DraftEncoder without oracle-text card features never beats the linear
   pool model, and with the deck it lands just under `linear_deck`, after 20-60 minutes of training instead of seconds.
   Card text, which this reimplementation doesn't have, is probably where the paper's gain comes from.
5. **Card value ≈ a single number per card.** Pair synergy is about 50× smaller than card power and does not improve
   held-out games. Learned weights track public GIH WR closely (Spearman 0.81). They are still worth fitting:
   on the same final pool, the GIH WR heuristic averages 0.18 across sets, against 0.21 for learned weights.
6. **Two traps:** win rates from cards *cast* look great per game (AUC 0.84) but score decks badly, because winners
   cast more. And rank lowers p at a fixed deck because it sets the opponents, so never read it as skill.

# Part 2: Next-game model: win probability from this game's deck, the record so far, and skill

This notebook predicts **the next game** of a Premier Draft run, rather than the run as a whole. Three
reasons it has to be done one game at a time:

* **Opponents depend on the record.** If Arena pairs players by record, a 5-1 deck faces tougher decks than a
  0-0 deck, so the same 40 cards are worth a different win probability at different points in the run.
* **The deck can change between games.** About one run in five rebuilds part-way through. Each game here uses
  the deck that was registered *for that game*.
* **Some context exists only per game**: play/draw, mulligans, both players' current rank.

The model is the game-level model from `scripts/deck_synergy.py`, plus the record before the game and the
player's experience bucket:

P(win game) = σ( b + f(deck) + record<sub>W-L</sub> + rank + opp rank + skill + experience + play/draw + mulligans )

f(deck) is either linear (card power) or a factorization machine (card power + pair synergy).

Sections:
1. How this differs from `decks.ipynb`
2. Data: one row per game, its own deck
3. Rebuilding the record before each game
4. Fitting the models
5. What the record does at a fixed deck
6. League (rank), skill and experience
7. Card power with and without the record, and pair synergy
8. Predicting the next game
9. Takeaways

Code: `src/decks/deckstrength.py` (`record_before`, `record_onehot`, `experience`, `covariates`) and
`src/decks/synergy.py` (the model). Needs `data/parquet/MSH.PremierDraft{,.game}.parquet`. The fits run on
CPU in a few minutes.

## 3. Rebuilding the record before each game

Game data has no "record so far" column, so it is rebuilt from the run's own games. Games are ordered by
`match_number`, and the record before a game counts the wins and losses in the run's earlier games.

17Lands does not have every game of every run, so the record is counted from **the games we have**. For a game
with earlier games missing, the record undercounts. That affects the 16% of games printed below; in most of them
only one earlier game is missing. The draft table's final W-L can't fill the gaps: it is a snapshot, and for many
runs the game file has *more* games than it records.

```python
print(f"games whose earlier games are all present: {known.mean():.1%}  ({known.sum():,} of {len(known):,}); "
      "all games are used")
gm, X, y, sp = games, X_all, y_all, split
wb, lb = wins_before, losses_before
print(f"train / val / test games: {np.sum(sp == 0):,} / {np.sum(sp == 1):,} / {np.sum(sp == 2):,}")

def record_grid(values, wins, losses, how="mean"):
    out = np.full((MAX_LOSSES, MAX_WINS), np.nan)
    n = np.zeros((MAX_LOSSES, MAX_WINS), dtype=int)
    for li in range(MAX_LOSSES):
        for wi in range(MAX_WINS):
            m = (wins == wi) & (losses == li)
            n[li, wi] = m.sum()
            if m.sum():
                out[li, wi] = values[m].mean() if how == "mean" else values[m][0]
    return out, n

def heat(ax, grid, n, title, fmt, cmap, vmin, vmax, center=None):
    im = ax.imshow(grid, cmap=cmap, vmin=vmin, vmax=vmax, aspect="auto")
    for (li, wi), v in np.ndenumerate(grid):
        if np.isfinite(v):
            ax.text(wi, li, fmt(v) + (f"\n{n[li, wi]:,}" if n is not None else ""), ha="center", va="center",
                    fontsize=7.5, color=INK)
    ax.set_xticks(range(MAX_WINS)); ax.set_yticks(range(MAX_LOSSES)); ax.grid(False)
    ax.set_xlabel("wins before the game"); ax.set_ylabel("losses before the game")
    ax.set_title(title, loc="left", fontsize=10)
    return im

raw, n_rec = record_grid(y.astype(float), wb, lb)
fig, ax = plt.subplots(figsize=(8.5, 3.4))
heat(ax, raw * 100, n_rec, "Raw game win rate by record before the game (%), and games", lambda v: f"{v:.1f}",
     "RdBu", 46, 66)
plt.tight_layout(); plt.show()
```

The raw win rate is almost **flat** across the record, between 55% and 57% everywhere. That is surprising. Only
strong decks and strong players get to 5-0 or 6-0, so if opponents were the same at every record, the win rate
would climb with wins. A flat line means opponents get tougher as the wins pile up, consistent with Arena pairing
players by record. Section 5 separates the two effects.

## 4. Fitting the models

Every model is a logistic regression in PyTorch (`synergy.fit`): Adam, L2 on the card weights, early stopping on
validation games. The regularisation is the value `deck_synergy.py` chose for MSH. The blocks of inputs:

* **deck**: card counts of this game's deck, lands included
* **game**: play/draw, user rank and opponent rank (one-hot, with "unknown"). All known before the game starts.
* **mulligans**: the user's mulligans minus the opponent's. Only known once the game has started, so the
  pre-game models leave it out; `deck_synergy.py` includes it, so its reproduction does too.
* **skill**: `user_game_win_rate_bucket`, the user's 17Lands game win rate. It is *leaky*: it is computed over all
  of the user's games in the data, including this one and later ones. Shown with and without.
* **experience**: `user_n_games_bucket` (1 / 5 / 10 / 50 / 100+ games logged). Much less leaky.
* **record**: one-hot of the record before the game, 0-0 to 6-2 (21 states)

```python
cov_full = covariates(gm).astype(np.float32)   # [play, mull, skill, skill_known, rank (7), opp rank (7)]
MULL, SKILL = [1], [2, 3]
game_cols = [i for i in range(cov_full.shape[1]) if i not in MULL + SKILL]
blocks = {"game": cov_full[:, game_cols], "mulligans": cov_full[:, MULL], "skill": cov_full[:, SKILL],
          "experience": experience(gm), "record": record_onehot(wb, lb)}

def context(*parts):
    """Concatenate blocks and remember where each one sits."""
    cols, at = [], {}
    for p in parts:
        at[p] = slice(sum(c.shape[1] for c in cols), sum(c.shape[1] for c in cols) + blocks[p].shape[1])
        cols.append(blocks[p])
    return np.hstack(cols), at

none = np.zeros_like(X[:, :1])
specs = [  # name, deck input, context blocks, FM rank k, l2 on card weights, l2 on pair vectors
    ("deck_synergy.py (+ mulligans)",   True,  ("game", "mulligans", "skill"), 0, 1e-4, 0),
    ("context only, no deck",           False, ("game", "skill", "experience", "record"), 0, 0, 0),
    ("deck + game + skill",             True,  ("game", "skill"), 0, 1e-4, 0),
    ("+ experience",                    True,  ("game", "skill", "experience"), 0, 1e-4, 0),
    ("+ experience + record",           True,  ("game", "skill", "experience", "record"), 0, 1e-4, 0),
    ("+ experience + record, no skill", True,  ("game", "experience", "record"), 0, 1e-4, 0),
    ("+ experience + record + synergy", True,  ("game", "skill", "experience", "record"), 4, 1e-5, 1e-3),
]
te_idx = np.flatnonzero(sp == 2)
models, preds, layout, rows = {}, {}, {}, []
for name, use_deck, parts, k, l2w, l2v in specs:
    t1 = time.time()
    C_in, at = context(*parts)
    m, val, predict = fit(X if use_deck else none, C_in, y, sp, k, l2w, l2v, log=lambda s: None)
    p = predict(te_idx)
    models[name], preds[name], layout[name] = m, p, at
    rows.append({"model": name, "val log loss": round(val, 5), "test log loss": round(log_loss(y[te_idx], p), 5),
                 "test AUC": round(roc_auc_score(y[te_idx], p), 4), "fit (s)": round(time.time() - t1)})
    print(f"{name:34s} test log loss {rows[-1]['test log loss']:.5f}  AUC {rows[-1]['test AUC']:.4f}  "
          f"[{rows[-1]['fit (s)']}s]", flush=True)
base_ll = log_loss(y[te_idx], np.full(len(te_idx), y[sp == 0].mean()))
res = pl.DataFrame(rows).with_columns((pl.col("test log loss") - base_ll).round(5).alias("vs constant"))
print(f"constant (train win rate {y[sp == 0].mean():.3f}): test log loss {base_ll:.5f}")
res
```

How to read the table (log loss is per test game, lower is better):

* **Mulligans are in-game information, and they are worth a lot.** `deck_synergy.py`'s model, which includes them,
  scores 0.6572. The same inputs without mulligans score 0.6679. A pre-game prediction can't know them, so every
  other row, and `predict_next` below, leaves them out.
* **The record helps on every metric**: -0.0014 log loss and +0.004 AUC over `+ experience`. For scale, the whole
  deck is worth 0.0044 (`context only, no deck` vs `+ experience + record`), so the record is worth about a third
  as much as knowing the 40 cards.
* **The skill bucket is the biggest single input**: dropping it costs 0.011, about 2.5× what the deck is worth.
  Part of that is leakage (it includes this game's result), so treat the rows with it as optimistic.
* **Experience adds almost nothing** once the skill bucket is in (0.66786 vs 0.66791). The two overlap.
* **Synergy doesn't help**: the FM is 0.0006 worse than the linear model on test games.

## 5. What the record does at a fixed deck

Left: the raw win rate from section 3. Right: the record coefficients of `+ experience + record`, relative to
0-0, converted to percentage points at the average win rate. This is the change in win probability for **the same
deck, player and context**, moving from 0-0 to that record.

```python
MAIN = "+ experience + record"
def block_weights(name, part):
    return models[name].cov.weight[0, layout[name][part]].detach().numpy()

rec_w = block_weights(MAIN, "record")
rec_w = rec_w - rec_w[0]
p0 = y[sp == 0].mean()
lo0 = np.log(p0 / (1 - p0))
rec_pts = (1 / (1 + np.exp(-(lo0 + rec_w))) - p0) * 100
adj = rec_pts.reshape(MAX_LOSSES, MAX_WINS)
adj[n_rec < 300] = np.nan  # too few games to read

fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 3.4))
heat(a1, raw * 100, None, "Raw win rate (%)", lambda v: f"{v:.1f}", "RdBu", 46, 66)
heat(a2, adj, None, "Same deck and player: change vs 0-0 (points)", lambda v: f"{v:+.1f}", "RdBu", -10, 10)
plt.tight_layout(); plt.show()
print("change vs 0-0, percentage points (rows = losses, cols = wins):")
print(np.round(adj, 1))

# deck strength by record: mean of the model's deck score for the decks that reach each record
deck_score = (X @ models[MAIN].w.detach().numpy())
ds_grid, _ = record_grid(deck_score, wb, lb)
print("mean deck score (log-odds, vs decks at 0-0) of the decks that reach each record:")
print(pl.DataFrame({f"{li} losses": np.round(ds_grid[li] - ds_grid[0, 0], 3) for li in range(MAX_LOSSES)})
      .with_columns(pl.Series("wins", range(MAX_WINS))).select("wins", *[f"{li} losses" for li in range(MAX_LOSSES)]))
```

At a fixed deck and player, **each win before the game costs about 2 points of win probability, and each loss gives
about 2.5-3 back**. At 6-0 the same deck wins 12.6 points less often than at 0-0. At 0-2 it wins 5.6 points more often.

The table underneath shows the other half of the story. The decks that reach 6-0 score 0.14 log-odds better than the
average deck at 0-0, about +3.5 points. Better decks reaching high records and tougher opponents at high records
cancel out, which is why the raw win rate is flat.

The record term cannot fully separate matchmaking from what the record says about the player. Being 5-0 also
means the deck and player are probably better than the model's inputs show, and that pushes the 5-0 coefficient
*up*. So the true matchmaking penalty is, if anything, larger than shown. Momentum or tilt would push the other way
(winners win more), and the data show the opposite.

For a run-level rating, this effect is already baked into the W-L target. For **the next game**, it has to be an input.

The other context coefficients of the same model. Ranks are relative to Bronze, experience relative to "1 game",

```python
gw = block_weights(MAIN, "game")
mull_w = block_weights("deck_synergy.py (+ mulligans)", "mulligans")[0]
rank_labels = ["unknown"] + list(RANK_ORDER)
user_rank, opp_rank = gw[1:1 + N_RANKS], gw[1 + N_RANKS:1 + 2 * N_RANKS]
exp_w = block_weights(MAIN, "experience")
skill_w = block_weights(MAIN, "skill")

fig, axes = plt.subplots(1, 3, figsize=(13.5, 3.4), gridspec_kw={"width_ratios": [1.6, 1.1, 0.9]})
xx = np.arange(N_RANKS)
axes[0].bar(xx - 0.2, user_rank - user_rank[1], width=0.4, color=BLUE, label="user's rank")
axes[0].bar(xx + 0.2, opp_rank - opp_rank[1], width=0.4, color=ORANGE, label="opponent's rank")
axes[0].set_xticks(xx, rank_labels, fontsize=8.5); axes[0].axhline(0, color=MUTED, lw=1)
axes[0].set_ylabel("log-odds vs Bronze"); axes[0].legend(fontsize=8.5); axes[0].grid(axis="x", visible=False)
axes[0].set_title("Rank (league)", loc="left")
axes[1].bar(range(len(exp_w)), exp_w - exp_w[1], color=AQUA, width=0.6)
axes[1].set_xticks(range(len(exp_w)), ["unknown", *N_GAMES_BUCKETS], fontsize=8.5)
axes[1].axhline(0, color=MUTED, lw=1); axes[1].grid(axis="x", visible=False)
axes[1].set_xlabel("games logged on 17Lands"); axes[1].set_title("Experience, vs 1 game", loc="left")
axes[2].bar([0], [skill_w[0] * 0.1], color=VIOLET, width=0.5)
axes[2].set_xticks([0], ["+10 pts win-rate\nbucket"]); axes[2].axhline(0, color=MUTED, lw=1)
axes[2].grid(axis="x", visible=False); axes[2].set_title("Skill (leaky)", loc="left")
for a in axes:
    a.set_ylabel("log-odds") if a is not axes[0] else None
plt.tight_layout(); plt.show()
print(f"on the play vs the draw: {gw[0]:+.3f} log-odds; each extra mulligan vs the opponent "
      f"(deck_synergy.py model): {mull_w:+.3f}")
```

* **Play/draw**: being on the play is worth +0.21 log-odds, about 5 points.
* **League**: the user's own rank is the bigger term. Higher rank means tougher opponents, from -0.09 at Silver to
  -0.39 at Diamond relative to Bronze. Arena pairs within rank, so the user's rank mostly says who they face. The
  opponent's rank adds a smaller penalty on top (about -0.03 to -0.18). "Unknown" rank is missing data, not a league.
* **Skill**: +0.39 log-odds per 10 points of the user's 17Lands win-rate bucket, the strongest context term. It is
  leaky, though. For live use, swap in the player's own win rate from *earlier* events only.
* **Experience**: players with more logged games win more, but the skill bucket already captures it (section 4).

## 7. Card power with and without the record, and pair synergy

Adding the record changes the card weights a little: part of what looked like card power was the record
(stronger decks get further, where opponents are tougher, so without the record their cards are
*under*-credited). Cards in at least 2,000 training games:

```python
tr = sp == 0
n_games_card = (X[tr] > 0).sum(0)
common = n_games_card >= 2000
w_no = models["+ experience"].w.detach().numpy()
w_rec = models[MAIN].w.detach().numpy()
w_noskill = models["+ experience + record, no skill"].w.detach().numpy()
fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 4.4))
for ax, other, lab in ((a1, w_no, "without the record"), (a2, w_noskill, "with the record, no skill bucket")):
    ax.scatter(other[common], w_rec[common], s=10, color=BLUE, alpha=0.6, lw=0)
    lim = [min(other[common].min(), w_rec[common].min()), max(other[common].max(), w_rec[common].max())]
    ax.plot(lim, lim, color=GRAY, ls="--", lw=1)
    slope = np.polyfit(other[common], w_rec[common], 1)[0]
    ax.set_xlabel(f"card weight, {lab}"); ax.set_ylabel("card weight, full model")
    ax.set_title(f"r = {np.corrcoef(other[common], w_rec[common])[0, 1]:.3f}, slope {slope:.2f}", loc="left")
plt.tight_layout(); plt.show()
power = (pl.DataFrame({"card": names, "weight (full)": w_rec, "weight (no record)": w_no,
                       "weight (no skill)": w_noskill, "train games": n_games_card})
         .filter(pl.col("train games") >= 2000).with_columns(pl.col(pl.Float32).round(3)))
print("Highest card power (log-odds per copy, full model)")
display(power.sort("weight (full)", descending=True).head(10))
```

**Adding the record barely moves the card weights** (r = 0.994). It stretches them by about 8%: without the record,
cards in strong decks were partly blamed for the tougher opponents those decks get paired against. Dropping the skill
bucket moves them more (r = 0.94). Without it, cards that good players take get some credit for the player.

**Pair synergy.** The FM adds a vector v per card, and a pair's synergy is ⟨v<sub>i</sub>, v<sub>j</sub>⟩ for
every pair of cards in the deck. Pairs that appear together in at least 2,000 training games:

```python
SYN = "+ experience + record + synergy"
V = models[SYN].v.detach().numpy()
P = (X[tr] > 0).astype(np.float32)
co = P.T @ P
iu = np.triu_indices(len(names), 1)
ok = co[iu] >= 2000
pairs = (pl.DataFrame({"card a": names[iu[0]][ok], "card b": names[iu[1]][ok], "synergy": (V @ V.T)[iu][ok],
                       "games together": co[iu][ok].astype(int)})
         .with_columns(pl.col("synergy").round(4)))
print(f"sd of pair synergy {pairs['synergy'].std():.4f} log-odds; sd of card power {w_rec[common].std():.3f}")
print("Strongest positive pairs"); display(pairs.sort("synergy", descending=True).head(8))
print("Strongest negative pairs"); display(pairs.sort("synergy").head(8))
```

## 8. Predicting the next game

`predict_next` takes a deck (card name → copies) and the situation before the game, and returns the win
probability. It builds the inputs in the same order as `covariates`, `experience` and `record_onehot`. Unknown
skill uses the training mean with the "known" flag off, as in training.

```python
model = models[MAIN]
idx_of = {n: i for i, n in enumerate(names)}
skill_mean = gm["user_game_win_rate_bucket"].cast(pl.Float32, strict=False).mean()

def predict_next(deck: dict, wins: int, losses: int, on_play: bool, rank: str | None = None,
                 opp_rank: str | None = None, skill: float | None = None,
                 n_games: str | None = None) -> float:
    x = np.zeros((1, len(names)), dtype=np.float32)
    for card, copies in deck.items():
        x[0, idx_of[card]] = copies
    onehot = lambda r: np.eye(N_RANKS)[0 if r is None else RANK_ORDER[r] + 1]
    game = np.r_[float(on_play) - 0.5, onehot(rank), onehot(opp_rank)]
    sk = np.r_[(skill_mean if skill is None else skill) - 0.5, float(skill is not None)]
    ex = experience(pl.DataFrame({"user_n_games_bucket": [n_games]}, schema={"user_n_games_bucket": pl.Utf8}))[0]
    c = np.r_[game, sk, ex, record_onehot(np.array([wins]), np.array([losses]))[0]].astype(np.float32)[None]
    with torch.no_grad():
        return float(torch.sigmoid(model(torch.from_numpy(x), torch.from_numpy(c))))

# sanity check: the function reproduces the model's own predictions on a few test games
check = te_idx[:5]
row = lambda i: gm.row(int(i), named=True)
def from_row(i):
    r = row(i)
    deck = {names[j]: int(X[i, j]) for j in np.flatnonzero(X[i])}
    rk = lambda v: v if v in RANK_ORDER else None
    sk = r["user_game_win_rate_bucket"]
    return predict_next(deck, int(wb[i]), int(lb[i]), r["on_play"] == "True", rk(r["rank"]), rk(r["opp_rank"]),
                        None if sk is None else float(sk), r["user_n_games_bucket"])
print(np.round([from_row(i) for i in check], 4), np.round(preds[MAIN][:5], 4))
```

**A test run, game by game.** A test draft that rebuilt its deck part-way through, with every game present.
Each row is the prediction made *before* that game: this game's deck, the record so far, play/draw, both ranks,
and the player's skill and experience buckets.

```python
ev = (gm.with_row_index("i").filter(pl.Series(sp == 2))
      .group_by("draft_id").agg(pl.col("build_index").n_unique().alias("builds"), pl.len().alias("n"))
      .filter((pl.col("builds") > 1) & (pl.col("n") >= 7)).sort("draft_id").row(0, named=True))
rows_ev = gm.with_row_index("i").filter(pl.col("draft_id") == ev["draft_id"]).sort(pl.col("match_number").cast(pl.Int32))
ii = rows_ev["i"].to_numpy()
prev = None
run = []
for i in ii:
    changed = 0 if prev is None else int(np.abs(X[i] - X[prev]).sum() / 2)
    run.append({"game": int(rows_ev.filter(pl.col("i") == i)["match_number"][0]),
                "record before": f"{wb[i]}-{lb[i]}", "build": int(gm["build_index"][int(i)]),
                "cards swapped": changed, "on play": gm["on_play"][int(i)] == "True",
                "rank": gm["rank"][int(i)], "opp rank": gm["opp_rank"][int(i)],
                "p(win)": round(from_row(i), 3), "won": bool(y[i])})
    prev = i
pl.DataFrame(run)
```

The predictions drift down as the wins pile up: that is the record term. The jumps between neighbouring games come
from the rebuild, play/draw and the opponent's rank. A new `build_index` doesn't always mean new cards: 17Lands
also starts a new build when only the sideboard changes.

**What-if: one deck across the whole run.** The same deck (the highest-scoring test deck) and an average
player at Gold, against a Gold opponent, at every record, on the play and on the draw.

```python
best = te_idx[np.argmax(deck_score[te_idx])]
deck_best = {names[j]: int(X[best, j]) for j in np.flatnonzero(X[best])}
grid_play = np.array([[predict_next(deck_best, w_, l_, True, "gold", "gold") for w_ in range(MAX_WINS)]
                      for l_ in range(MAX_LOSSES)])
grid_draw = np.array([[predict_next(deck_best, w_, l_, False, "gold", "gold") for w_ in range(MAX_WINS)]
                      for l_ in range(MAX_LOSSES)])
fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 3.2))
lo_, hi_ = min(grid_play.min(), grid_draw.min()) * 100 - 1, max(grid_play.max(), grid_draw.max()) * 100 + 1
heat(a1, grid_play * 100, None, "On the play: p(win) %", lambda v: f"{v:.1f}", "RdBu", lo_, hi_)
heat(a2, grid_draw * 100, None, "On the draw: p(win) %", lambda v: f"{v:.1f}", "RdBu", lo_, hi_)
plt.tight_layout(); plt.show()
```

## 9. Takeaways

1. **Model the next game, not the run.** One run in five changes its deck part-way through, and each game here uses
   the deck registered for it. The model takes the deck, the record so far, both ranks and play/draw, and returns a
   pre-game win probability (`predict_next`).
2. **The record matters, and it is matchmaking.** At a fixed deck and player, each win costs about 2 points of win
   probability and each loss gives back about 2.5. Better decks reaching high records hide this in the raw data.
   Adding the record improves held-out log loss by 0.0014, about a third of what the whole deck adds.
3. **League and skill.** The user's own rank is mostly about who they face (higher rank, lower win probability at a
   fixed deck). The 17Lands skill bucket is the strongest input but leaks the outcome; a live version should use
   win rate from earlier events only. Experience adds nothing on top of skill.
4. **Card power is stable, synergy is small.** Adding the record changes card weights by about 8% in scale and
   almost nothing in order. Pair synergy is about 40× smaller than card power and doesn't improve held-out games.
5. **Mulligans are the biggest in-game signal** (-0.63 log-odds per extra mulligan), but only for in-game or
   post-hoc models.
6. **Sealed** fits the same pipeline: the game file has the same deck and record columns, but no rank. There, the
   record, skill and draft-trained card weights (multi-task) will have to carry more of the load.
